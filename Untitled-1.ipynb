```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file_path = "/Users/roger/Desktop/UChicago/Research/PME_F/Dataset_Analysis/flow_and_cytokine_data_trimmed.csv"
df = pd.read_csv(file_path)
df.head()

```

```python
agonist = ["Pam3", "Pam2", "R848", "cGAMP", "Fla", "Poly IC", "PAM3", "PAM2", "LPS"]
agonist_R848 = [name for name in df.columns if (name.startswith("R848") and name.startswith("R848 MFI") == False)] 
agonist_cGAMP = [name for name in df.columns if (name.startswith("cGAMP") and name.startswith("cGAMP MFI") == False)]

log_transformed_R848 = np.log1p(df[agonist_R848])
log_transformed_cGAMP = np.log1p(df[agonist_cGAMP])
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA


scaler = StandardScaler()
standardized_R848 = scaler.fit_transform(log_transformed_R848)
standardized_cGAMP = scaler.fit_transform(log_transformed_cGAMP)

pca = PCA(n_components=2)
reduced_R848 = pca.fit_transform(standardized_R848)
reduced_cGAMP = pca.fit_transform(standardized_cGAMP)

sns.set_theme(style="white")
sns.scatterplot(x=reduced_R848[:, 0], y=reduced_R848[:, 1])
sns.scatterplot(x=reduced_cGAMP[:, 0], y=reduced_cGAMP[:, 1])

```

```python
Amodiaquine = reduced_R848[478, :].reshape(1, -1)
Amodiaquine
```

```python
from sklearn.neighbors import NearestNeighbors

KNN = NearestNeighbors(n_neighbors=10).fit(reduced_R848)

distance, indices = KNN.kneighbors(X=Amodiaquine)
indices.reshape(10)

#df[df["Name"] == "Amodiaquine dihydrochloride dihydrate"]

Amodiaquine_NN_R848 = df.iloc[indices.reshape(10)]["Name"]
```

```python
sns.scatterplot(x=reduced_R848[:, 0], y=reduced_R848[:, 1])
sns.scatterplot(x=Amodiaquine[:, 0], y=Amodiaquine[:, 1])
for i, name in zip(indices.reshape(10), Amodiaquine_NN_R848):
    plt.scatter(x=reduced_R848[i, 0], y=reduced_R848[i, 1], label=name)
```
